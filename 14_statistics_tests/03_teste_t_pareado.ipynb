```python
from scipy.stats import ttest_rel
import numpy as np
```

```python
mae_rna = [45.5012, 52.3862, 39.7903, 30.8527, 27.1338]
mae_rl_polynomial = [29.8763, 29.1873, 28.9052, 30.8245, 31.0714]
```

```python
t_stat, p_value = ttest_rel(mae_rna, mae_rl_polynomial)
```

```python
print(f"Estatística t: {t_stat:.4f}")
print(f"Valor-p: {p_value:.4f}")

# Interpretando o resultado
alpha = 0.05  # Nível de significância de 5%
if p_value < alpha:
    print("Rejeitamos a hipótese nula: Os modelos têm desempenhos significativamente diferentes.")
else:
    print("Não rejeitamos a hipótese nula: Não há evidência de diferença significativa entre os modelos.")
```

```python
# Cálculo da média do MAE para cada modelo
mean_mae_rna = np.mean(mae_rna)
mean_mae_rl_polynomial = np.mean(mae_rl_polynomial)

print(f"Média MAE RNA: {mean_mae_rna:.4f}")
print(f"Média MAE RL Polynomial: {mean_mae_rl_polynomial:.4f}")

if mean_mae_rna < mean_mae_rl_polynomial:
    print("A RNA tem menor MAE médio, portanto, é melhor em termos de erro absoluto médio.")
else:
    print("A Regressão Polinomial tem menor MAE médio, portanto, é melhor em termos de erro absoluto médio.")
```
