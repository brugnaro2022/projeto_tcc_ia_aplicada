<h4><b>Detecção de outliers</b></h4> 
<p>Valores anormais</p>
<p>Valores fora do padrão (afastados da média)</p>

<p><b>Causas</b></p>
<ul>
    <li>Acaso</li>
    <li>Erro no preenchimento dos dados</li>
    <li>Fraudes</li>
</ul>

<p><b>Tratamento</b></p>
<ul>
    <li>Remover o registro</li>
    <li>Não fazer nada</li>
    <li>Substituir</li>
</ul>

<p><b>Percorremos uma lista com todos os dados das features numéricas, e foram verificadas algumas features que contém outliers:</b></p>
<ul>
    <li>line_category</li>    
    <li>line_color</li>    
    <li>accessibility_info_eletric_ramp</li>    
    <li>station_initial_latitude</li>    
    <li>station_final_latitude</li>
    <li>distance</li>    
    <li>distance_km</li>
    <li>average_speed_ms</li>    
    <li>average_speed_kh</li>
    <li>duration_min</li>    
    <li>duration</li>
</ul>

```python
df_numeric = df.select_dtypes(include=['float64', 'int64'])

outliers_sums = {}

for index, feature in enumerate(df_numeric):
    Q1 = df[feature].quantile(0.25)
    Q3 = df[feature].quantile(0.75)
    IQR = Q3 - Q1
    
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    
    idx = np.where((df[feature] < limite_inferior) | (df[feature] > limite_superior))

    outliers = df[feature].iloc[idx]
    
    sum_outliers = outliers.count()

    if sum_outliers != 0:
        print(f'feature: {feature}, soma dos outliers: {sum_outliers}')
```

<h4><b>Matriz de Correlação</b></h4>

<p>Exibindo os valores da matrix de correlação completa</p>

```python
correlation_matrix = df_numeric.corr()
correlation_matrix
```

<p>Plotando gráfico da matrix de correlação</p>

```python
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', square=True, cbar_kws={"shrink": .8})
plt.title('Correlation Heatmap')
plt.show()
```

<p>Ranqueando as 20 melhores combinações de features com correlações consideráveis</p>

```python
corr_pairs = correlation_matrix.unstack().reset_index()
corr_pairs.columns = ['Feature1', 'Feature2', 'Correlation']

corr_pairs = corr_pairs[corr_pairs['Feature1'] != corr_pairs['Feature2']]

corr_pairs = corr_pairs[corr_pairs['Correlation'] > 0]

corr_pairs['Pair'] = corr_pairs.apply(lambda x: tuple(sorted((x['Feature1'], x['Feature2']))), axis=1)

unique_corr_pairs = corr_pairs.drop_duplicates(subset='Pair').reset_index(drop=True)

sorted_corr = unique_corr_pairs.sort_values(by='Correlation', ascending=False)

top_n = 20
best_corr_features = sorted_corr.head(top_n)

best_corr_features[['Feature1', 'Feature2', 'Correlation']]
```

<p>Plotando gráfico de barras visando melhorar a visualização das 20 melhores combinações de features correlacionadas</p>
